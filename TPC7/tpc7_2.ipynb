#### Função que dada uma string e uma substring não vazia, calcula  o número de vezes em que a substring aparece na string

```python
def strCount(s, subs):
    res = 0
    i = 0
    while i <= len(s) - len(subs):
        if s[i:i + len(subs)] == subs:
            res = res + 1
            i = i + len(subs)  
        else:
            i = i + 1
    return res


print(strCount("catcowcat", "cat"))
print(strCount("catcowcat", "cow"))
print(strCount("catcowcat", "dog"))
```

#### Função que recebe uma lista de números inteiros positivos e devolve o menor produto que for possível calcular multiplicando os 3 menores inteiros da lista:

```python

def produtoM3(lista):
    menores = []
    i = 0
    while i < 3:
        menor = lista[0]
        for elem in lista:
            if elem < menor:
                menor = elem
        menores.append(menor)
        lista.remove(menor)
        i = i + 1

    return menores[0] * menores[1] * menores[2]

print(produtoM3([12,3,7,10,12,8,9]))
```

#### Função que dado um número inteiro positivo, repetidamente adiciona os seus dígitos até obter-se apenas um dígito

```python

def reduxInt(n):
    while n>=10:
        soma = 0
        for car in str(n):
            soma = soma + int(car)
        n = soma
    
    return n

print(reduxInt(9846)) 
```
