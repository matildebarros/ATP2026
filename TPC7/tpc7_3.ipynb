Cada dicionário, correspondente a um _post_ e tem chaves `id`, `conteudo`, `autor`, `dataCriacao` e `comentarios`.
Por sua vez, `comentarios` é uma lista de dicionários com chaves `comentario` e `autor`.

```python
MyFaceBook = [{
    'id': 'p1', 
    'conteudo': 'A tarefa de avaliação é talvez a mais ingrata das tarefas que um professor tem de realizar...', 
    'autor': 'jcr', 
    'dataCriacao': '2023-07-20', 
    'comentarios': [
        {
            'comentario': 'Completamente de acordo...',
            'autor': 'prh'
        },
        {
            'comentario': 'Mas há quem goste...',
            'autor': 'jj'
        }
    ]},
    {
    'id': 'p2',
    'conteudo': 'informacao sobre conteudo',
    'autor': 'joao',
    'dataCriacao': '2025-05-24',
    'comentarios': [
        {
            'comentario': 'eu concordo com...',
            'autor': 'joana'
        }
    ]},
    ]
```

### Função que indica quantos posts estão registrados

```python
def quantosPost(redeSocial):
    return len(redeSocial) 

print(quantosPost(MyFaceBook))
```

### Função que devolve a lista de posts de um determinado autor

```python


def postsAutor(redeSocial, autor):
    i = 0
    res = []

    while i < len(redeSocial):
        if redeSocial[i]["autor"] == autor:
            res.append(redeSocial[i])
        i = i + 1

    return res


print(postsAutor(MyFaceBook, "jcr"))
```

### Função que devolve a lista de autores de posts ordenada alfabeticamente

```python
def autores(redeSocial):
    i = 0
    autores = []

    while i < len(redeSocial):
        autores.append(redeSocial[i]["autor"])
        i = i + 1
    
    
    trocou = True
    while trocou == True:
        trocou = False
        j = 0
        while j < len(autores) - 1:
            if autores[j]>=autores[j+1]:
                n = autores[j]
                autores[j] = autores [j+1]
                autores[j+1] = n
                trocou = True
            j = j + 1

    
    return  autores

print(autores(MyFaceBook))
```

### Função que acrescenta um novo post à rede social a partir dos parâmetros recebidos


```python
def insPost(redeSocial, conteudo, autor, dataCriacao, comentarios):
    id = "p" + str(len (redeSocial) + 1)

    post = {}
    post["id"] = id
    post["conteudo"] = conteudo
    post["autor"] = autor
    post["dataCriacao"] = dataCriacao
    post["comentarios"] = comentarios

    redeSocial.append(post)
    return redeSocial

print(insPost(MyFaceBook,"conteudo", "catarina","2026-10-08","comentario"))
```

### Função que remove um post da rede, correspondente ao `id` recebido.

```python
def remPost(redeSocial, id):
    i = 0
    while i < len(redeSocial):
        if redeSocial[i]["id"] == id:
            del redeSocial[i]
        else:
            i = i + 1
    return redeSocial

print(remPost(MyFaceBook, "p3"))
```

### Função que devolve uma distribuição de posts por autor

```python
def postsPorAutor(redeSocial):
    dist = {}
    i = 0
    while i < len(redeSocial):
        autor = redeSocial[i]["autor"]
        if autor in dist:
            dist[autor] = dist[autor] + 1
        else:
            dist[autor] = 1
        i = i + 1

    return dist

print(postsPorAutor(MyFaceBook))
```

### Função que recebe um autor e devolve a lista de posts comentados por esse autor

```python
def comentadoPor(redeSocial, autor):
    i = 0
    res = []
    while i < len(redeSocial):
        j = 0
        encontrou = False

        while j < len(redeSocial[i]["comentarios"]) and not encontrou:
            if redeSocial[i]["comentarios"][j]["autor"] == autor:
                res.append(redeSocial[i])
                encontrou = True
            j = j + 1
        
        i = i + 1
    return res

print(comentadoPor(MyFaceBook,"jj"))
```
